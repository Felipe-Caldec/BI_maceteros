```python
import pandas as pd
from datetime import datetime
from dotenv import load_dotenv
import os

load_dotenv()

csv_path_nuevo = os.environ["CSV_PATH_DF_NUEVO"]
csv_path_antiguo = os.environ["CSV_PATH_DF_ANTIGUO"]

fecha = datetime.now().strftime("%d-%m-%Y")
df_nuevo = pd.read_csv(csv_path_nuevo, sep=",", encoding="utf-8")
df_antiguo = pd.read_csv(csv_path_antiguo, sep=",", encoding="utf-8")

df = pd.concat([df_nuevo,df_antiguo], axis=0, ignore_index=True)

df['Precio'] = df["Precio"].astype("Int64")
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]
for col in df.columns:
    print(repr(col))

df.to_excel(f"consolidado_final_{fecha}.xlsx")
```
