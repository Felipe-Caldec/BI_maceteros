```python
import pandas as pd
import re
from dotenv import load_dotenv
import os

load_dotenv()
path_productos_amazon = os.environ['PATH_PRODUCTOS_AMAZON']

df= pd.read_csv(path_productos_amazon, sep=",", encoding="utf-8")
print(df.head())

df["precio_clp"] = df["Precio envio"].str.extract(r"CLP\s*([\d.,]+)")
df["precio_clp"] = df["precio_clp"].str.replace(",","").str.replace('.','') # con str hace que reemplaza cualquier caracter de la cadena, sino busca coincidencia exaxta

df["precio_clp"] =( pd.to_numeric(df["precio_clp"], errors="coerce").astype("Int64"))
df["Precio"] = (pd.to_numeric(df["Precio"], errors="coerce").astype("Int64"))

print(df[["precio_clp","Precio"]].dtypes)

df['Precio'] = df["precio_clp"] + df['Precio']
df = df.drop(columns=['precio_clp', 'Precio envio'])

print(df.head())
#print(df_amazon["precio_clp"].head())
```

```python
df['Nombre'] = df['Nombre'].str.upper()
#df['Marca'] = df['Marca'].str.upper()
#df['Marca'] = df['Marca'].replace("GENÉRICO", 'GENERICO').replace("GENERICA","GENERICO")

print(df.head())
```

```python
import re
# variable medida
def extraer_medida(texto):
    patron = r"\d+\s*x\s*\d+\s*cm|\d+\s*cm"
    match = re.search(patron, texto, re.IGNORECASE)
    if match:
        return match.group().replace(" ", "")
    return None

df["Medidas"] = df["Nombre"].apply(extraer_medida)

print(df.head())
```
